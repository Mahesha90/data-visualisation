Bar chart for years of practice per athleate

pyplot is a submodule for matplotlib
plt is a comman alias for pyplot. 

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Years of practice per athlete")
plt.ylabel("Years")
plt.show()
```

Bar chart for age per athlete


```python
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age of Athletes")
plt.ylabel("Age")
plt.show()
```

Bubble chart for age vs years of practice

plt.scatter (x,y) makes a point for each pair of value
x=age, y=years of practice
s=controls the size of each point (bubble). If the student has more years of practice the bubble is lerger.
alpha 0.6 makes the bubble transparent. So it will shows bubble is there is any overlaps.


```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Yeas=rs of practice")
plt.title("Age vs years of practice per athlete")
plt.show()
```

Box plot for years of practice and age

```python
import pandas as pd
import matplotlib.pyplot as plt

#df = pd.read_csv("sports.csv")
#df

#calculate quartiles and extremes
#Age
age_min = df["Age"].min()
age_q1 = df["Age"].quantile(0.25)
age_q2 = df["Age"].quantile(0.5)
age_q3 = df["Age"].quantile(0.75)
age_max = df["Age"].max()

#Years of practice
years_min = df["Years_of_practice"].min()
years_q1 = df["Years_of_practice"].quantile(0.25)
years_q3 = df["Years_of_practice"].quantile(0.75)
years_max = df["Years_of_practice"].max()

print("Min:", age_min, "Q1:", age_q1, "Q2:", age_q2, "Q3:", age_q3, "Max:", age_max)
print("Min:", years_min, "Q1:", years_q1, "Q3:", years_q3, "Max:", years_max)

#Draw box plot
plt.figure(figsize=(8,5))
df[["Age", "Years_of_practice"]].boxplot()
plt.title("Box plot for age and years of practice")
plt.show
```

```python
avg_age = df.groupby("Sport")["Age"].mean()
avg_age.plot(kind="bar")
plt.ylabel("average age")
plt.title("Bar chart for Average age per sport")
plt.show
```
