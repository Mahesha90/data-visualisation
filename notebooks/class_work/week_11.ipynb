```python
sentence = "racecarlevelfield" #letter count always starts from 0
print("First three letters:", sentence[0:3]) #extracting 3 first letters
print ("Extracted letters 5-9:", sentence[4:9])
print ("Extract letters starting from 5th til end:", sentence[4:])
print ("Last character:", sentence[-1])
print ("Last 4 characters :", sentence[-4:])
print ("Substring starting from 5:", sentence[5:][0]) #extracts a substring starting from 5, the 0 in brackets will take the first character of that substring, therefore only printing 1 letter
print ("Substring starting from 5 + 5 letters:", sentence[5:][:5])
print (sentence)

#extracting every second character
print (sentence[::2]) #the number here shows,  that it's second character, replacing the number with 3 would take every 3rd letter
```

```python
sentence2 = "Alice went to the park and saw a beautiful bird singing"

words = sentence2.split(" ") #splitting the sentence into words, by identifying spaces inbetween words
print (words)
print ("Printing only first word identified:", words [0])
print ("Printing first 3 words: ", *words [:3])
print ("Printing last 3 words:", *words[-3:])
print ("Number of words:", len (words))
#savesave
```
