```python
sentence = "flowers are green blue and yellow" #letter count always starts from 0
print("First six letters:", sentence[0:6]) #extracting 6 first letters
print ("Extracted letters 12-14:", sentence[11:14])
print ("Extract letters starting from 8th til end:", sentence[7:])
print ("Last character:", sentence[-1])
print ("Full sentence:", sentence)

#extracting every second character
print ("Every third charatcer:", sentence[::3]) #the number here shows after which characther another letter is printed

words = sentence.split(" ") #splitting the sentence into words, by identifying spaces inbetween words
print (words)
print ("Number of words:", len (words))
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt #Import for showing the plot/image

#Read the text file that I've added to the space
with open("HW text - week 11.txt", "r", encoding="utf-8") as file:
    text = file.read()

#Create and generate the word cloud object
wc = WordCloud(
    width=800, 
    height=400, 
    background_color='white',
    max_words=100
).generate(text)

#Save the image to a file
wc.to_file("alice_wordcloud.png")

#Displaying the image after running the code
plt.figure(figsize=(10, 5))
plt.imshow(wc, interpolation='bilinear')
plt.axis("off")
plt.show()
```

Considering that the text I've used, is a snippet from Alice in Wonderland, the word results are pretty valid.
There's a lot of usage of ''Alice'' and ''Rabbit, and many verbs such as: sitting, picking, sitting
Additionally the author is really descripitive, therefore many adjectives: pink, white, stupid, close
