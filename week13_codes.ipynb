```python
!pip install stanza
import stanza

stanza.download('en')
nlp = stanza.Pipeline('en')
```

```python
# Write  a program to show all the nouns from the text (User can enter the text inside the text box)

text_area = widgets.Textarea(description="Write text")

answer_label = widgets.Label(value="The answer will appear here")

button1 = widgets.Button(description="Press")
display(text_area,button1, answer_label)

def on_click(button):
  document = nlp(text_area.value)

  nouns = [word.text for sentence in document.sentences for word in sentence.words if word.xpos == "NN"]

  answer_label.value = "Nouns: " + ", ".join(nouns)

button1.on_click(on_click)


```

```python

text_area = widgets.Textarea(description="Write text")

answer_label = widgets.Label(value="The answer will appear here")

button_noun = widgets.Button(description="Nouns")
button_verbs = widgets.Button(description="Verbs")


def on_click(button):

  if button.description  == "Nouns":
    pos_tag = ["NN", "NNS", "NNP", "NNPS"]
  elif button.description == "Verbs":
    pos_tag = ["VB", "VBD", "VBG", "VBN", "VBP", "VBZ"]

  document = nlp(text_area.value)

  words_found = [word.text for sentence in document.sentences for word in sentence.words if word.xpos in pos_tag]

  answer_label.value = button.description + ":" + "," .join(words_found)

button_noun.on_click(on_click)
button_verbs.on_click(on_click)

display(text_area, button_noun, button_verbs, answer_label)
```
